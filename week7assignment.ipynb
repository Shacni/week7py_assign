```python
# Convert Gender to category type
df['Gender'] = df['Gender'].astype('category')

# Group by Gender and Academic Level instead of Country
gender_academic = df.groupby(['Gender', 'Academic Level']).size().reset_index(name='Count')

# Create a bar plot for Gender and Academic Level
plt.figure(figsize=(12, 6))
# Convert categorical to string before concatenation
plt.bar(gender_academic['Gender'].astype(str) + ' - ' + gender_academic['Academic Level'], 
        gender_academic['Count'], color='skyblue')
plt.title('Count by Gender and Academic Level')
plt.xlabel('Gender - Academic Level')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')  # Rotate labels for better readability
plt.grid(True, axis='y')
plt.tight_layout()
plt.show()
```
